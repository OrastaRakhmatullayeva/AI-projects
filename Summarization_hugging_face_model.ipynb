```python
import transformers

print(transformers.__version__)
```

```python
from transformers import pipeline
```

```python
summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)
```

```python
text = """
Artificial intelligence is changing the way people work and study.
Many companies use AI systems to automate repetitive tasks,
analyze large amounts of data, and improve decision-making.
In education, AI tools can help students understand difficult
topics and provide personalized learning experiences.
However, experts warn that AI systems can sometimes produce
incorrect information or reflect biases in their training data.
For this reason, responsible AI development requires careful
testing, human oversight, and transparency.
"""
```

```python
result=summarizer(
    text,
    max_length=30,
    min_length=10,
    do_sample=False
)
```

```python
print(result[0]["summary_text"])
```

```python
print(summarizer.task)
```

COMPARING MAX_LENGTH ANDD MIN_LENGTH

```python
short = summarizer(
    text,
    max_length=30,
    min_length=10,
    do_sample=False
)

medium = summarizer(
    text,
    max_length=50,
    min_length=20,
    do_sample=False
)

long = summarizer(
    text,
    max_length=80,
    min_length=30,
    do_sample=False
)

print("SHORT:", short[0]["summary_text"])
print("MEDIUM:", medium[0]["summary_text"])
print("LONG:", long[0]["summary_text"])
```

```python
def summarize(text):
    result = summarizer(
        text,
        max_length=60,
        min_length=20,
        do_sample=False
    )
    return result[0]["summary_text"]
```

```python
def summarize(text, max_length=60, min_length=20):
    result = summarizer(
        text,
        max_length=max_length,
        min_length=min_length,
        do_sample=False
    )

    return result[0]["summary_text"]
```

```python
print(summarize(text))
```

EVALUATION PART

```python
summary = summarize(text)

print("Original words:", len(text.split()))
print("Summary words:", len(summary.split()))
print("Summary:", summary)
```

### Conclusion

I used Hugging Face’s BART model to summarize a text and tested different `max_length` and `min_length` values. The model captured the main ideas well, but shorter summaries sometimes missed important details. This helped me understand how these settings affect the final summary.

BUILDING BEHIND PIPELINE MODEL:

PREPROCESSING -> MODEL -> POSTPROCESSING

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL = "facebook/bart-large-cnn"


tokenizer = AutoTokenizer.from_pretrained(MODEL)

model = AutoModelForSeq2SeqLM.from_pretrained(MODEL)

# 1. Preprocessing
inputs = tokenizer(
    text,
    return_tensors="pt",
    truncation=True,
    max_length=1024
)

# 2. Generate summary
outputs = model.generate(
    **inputs,
    max_length=60,
    min_length=20,
    do_sample=False,
    num_beams=4
)

# 3. Decode
summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(summary)
```
