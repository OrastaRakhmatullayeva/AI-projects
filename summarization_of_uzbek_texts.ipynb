```python
from transformers import pipeline

uz_summarizer = pipeline(
    "summarization",
    model="csebuetnlp/mT5_multilingual_XLSum"
)
```

```python
text_uz = """
Sun'iy intellekt bugungi kunda ta'lim, tibbiyot va sanoat
sohalarida keng qo'llanilmoqda. Ko'plab kompaniyalar
sun'iy intellekt yordamida katta hajmdagi ma'lumotlarni
tahlil qilmoqda va ish jarayonlarini avtomatlashtirmoqda.
Ta'lim sohasida esa AI o'quvchilarga murakkab mavzularni
tushunishga va individual ta'lim olishga yordam beradi.
Biroq sun'iy intellekt ba'zan noto'g'ri ma'lumotlar
taqdim etishi mumkin. Shu sababli mutaxassislar AI
tizimlarini doimiy tekshirish va mas'uliyat bilan
ishlatish muhimligini ta'kidlamoqda.
"""
```

```python
result = uz_summarizer(
    text_uz,
    max_length=80,
    min_length=20,
    do_sample=False
)

print(result[0]["summary_text"])
```

. Behind the Pipeline

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL = "csebuetnlp/mT5_multilingual_XLSum"

tokenizer = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL)

inputs = tokenizer(
    text_uz,
    return_tensors="pt",
    truncation=True,
    max_length=512
)

model.eval()

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_length=80,
        min_length=20,
        num_beams=4,
        do_sample=False
    )

summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(summary)
```
